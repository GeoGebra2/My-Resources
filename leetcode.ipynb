```python
import time
def find_prime_1(n):
    time_start = time.time()
    number_pn = 0

    for i in range(1, n):
        if is_prime_1(i):
            number_pn += 1

    time_end = time.time()
    return time_end - time_start, number_pn

def is_prime_1(n):
    if n == 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print(find_prime_1(10000))
```

```python
import time
def find_prime_2(n):
    time_start = time.time()
    number_pn = 0

    for i in range(1, n):
        if is_prime_2(i):
            number_pn += 1

    time_end = time.time()
    return time_end - time_start, number_pn

def is_prime_2(n):
    if n == 1:
        return False
    x = 2
    while x ** 2 <= n:
        if n % x == 0:
            return False
        x += 1
    return True

print(find_prime_2(1000000))
```

```python
import time
def find_prime_4(n):
    time_start = time.time()
    number_pn = 0
    num_lst = sieve(n)
    number_pn = [x for x in num_lst if x]
    time_end = time.time()
    return time_end - time_start, len(number_pn)

def sieve(n):
    if n <= 1:
        return []
    truth_lst = [True for x in range(n+1)]
    truth_lst[0], truth_lst[1] = False, False
    for i in range(2, n+1):
        if truth_lst[i] == False:
            continue
        for j in range(2*i, n+1, i):
            truth_lst[j] = False
    return truth_lst

print(find_prime_4(1000000))
```

```python
# 400-Nth Digit
def nth_digit(n):
    num_str = "1"
    num = 1
    while len(num_str) < n:
        num += 1
        num_str += str(num)
    num_lst = list(num_str)
    return num_lst[n-1]

print(nth_digit(3))
print(nth_digit(11))
```
