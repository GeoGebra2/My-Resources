```python
def is_happy_number(n):
    ans_lst = []
    while n != 1:
        sum = 0
        lst = list(str(n))
        for i in lst:
            sum += (int(i)) ** 2
        if sum in ans_lst:
            return False
        ans_lst.append(sum)
        n = sum
    
    return True

print(is_happy_number(19))
```

```python
def is_anagram(s,t):
    if len(s) != len(t):
        return False

    lsts = list(s)
    lstt = list(t)

    lsts.sort()
    lstt.sort()

    for i in range(len(lsts)):
        if lsts[i] != lstt[i]:
            return False

    return True

print(is_anagram("anagram", "nagaram"))
print(is_anagram("rat", "cat"))
```

```python
def insterrsection(list1, list2):
    ins_lst = []
    for i in list1:
        if i in list2:
            ins_lst.append(i)
            list2.remove(i)

    return ins_lst

print(insterrsection([1,2,2,1],[2,2]))
print(insterrsection([9,4,9,8,4], [4,9,5]))
```
