```python
import cvxpy as cp
```

```python
# T2解答
x1 = cp.Variable()
x2 = cp.Variable()

# Create two constraints.
constraints = [2*x1 + x2 >= 1,
               x1 + 4*x2 >= 1,
               x1 >= 0, x2 >= 0]

# Form objective.
obj = cp.Minimize(x1**2 + 8*x2**2)

# Form and solve problem.
prob = cp.Problem(obj, constraints)
prob.solve()  # Returns the optimal value.
print("status:", prob.status)
print("optimal value", prob.value)
print("optimal var", x1.value, x2.value)
```

```python
# T3原题解答
import cvxpy as cp
import numpy as np
m=3
n=2
A=[[3,1,1],[1,-3,2]]
b=[5,8,-5]

# Construct the problem.
x = cp.Variable(n)
objective = cp.Minimize(cp.norm1(A @ x - b))
constraints = [cp.norm_inf(x) <= 1]
prob = cp.Problem(objective, constraints)

print("Optimal value", prob.solve())
print("Optimal var")
print(x.value) # A numpy ndarray.
```

```python
# T4(b)、(c)
import cvxpy as cp
w = cp.Variable(2)
X = [[2,0,0],[0,1,0]]
y = [3,2,2]
# Create two constraints.
constraints = [(cp.norm2(w))**2 <= 100]

# Form objective.
obj = cp.Minimize(cp.norm2((X@w - y))**2)

# Form and solve problem.
prob = cp.Problem(obj, constraints)
prob.solve()  # Returns the optimal value.
print("status:", prob.status)
print("optimal value", prob.value)
print("optimal var", w.value)
```
