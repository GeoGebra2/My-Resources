```python
import numpy as np
```

```python
X = np.array([[2,0],[0,1],[0,0]])
y = np.array([3,2,2])
```

```python
print("X",X.shape,"y",y.shape)
```

```python
def fp(w): #计算梯度的函数
    return 2*(X.T@X@w - X.T@y)
def f(w): #目标函数
    return np.linalg.norm(X@w - y, ord=2)**2
```

```python
w = np.array([0.0,0.0]) #初始向量
w_traces = []
for i in range(1000):
    w_traces.append(w.copy()) #复制
    w_grad = fp(w)
    if(np.linalg.norm(w_grad) < 1e-5):
        break
    w -= w_grad * 0.01
print(w)
print(w_traces)
```
