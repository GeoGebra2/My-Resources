```python
import numpy as np
import copy
```

```python
def soft_th(w, th):
	#   START OF YOUR CODE
	b = []
	for i in range(len(w)):
		b.append(np.sign(w[i]) * np.max(abs(w[i]) - th, 0))
  
	return np.array(b).reshape(2,)
	#	END OF YOUR CODE

def ista(X, y, lambd, w0, stepsize, tol=1e-9, maxiter=10000):

	w_traces = [np.array(w0)]
	w = np.array(w0)

	for k in range(maxiter):
		#   START OF YOUR CODE
		a = w - stepsize * (X.T) @ (X @ w - y)
		w_next = soft_th(a, stepsize * lambd)
		#	END OF YOUR CODE

		if np.linalg.norm(w_next - w) < tol: break

		w = w_next
		w_traces.append(np.array(copy.deepcopy(w)))


	return w_traces 
```

```python
X = np.array([[2,0,0], [0,1,0]]).T
y = np.array([3,2,2])
lambd = 2

def f(w):
	return 0.5*np.sum((X@w - y)**2) + lambd * np.linalg.norm(w,ord=1)

w0 = 0.6 * np.ones(2)
stepsize = 0.1

w_traces = ista(X, y, lambd, w0, stepsize=stepsize, tol=1e-9)

f_value = f(w_traces[-1])
```
