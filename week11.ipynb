```python
def str_to_list(s):
    s = list(s)
    s.reverse()
    if s[-1] == "+":
       s.pop()
    return s

def list_to_str(lst):
    temp = [str(x) for x in lst[::-1]]
    return "".join(temp)

print(list_to_str([1,2,3,4,"-"]))
print(str_to_list("4"))
```

```python
def mul_ten(s, t, k):
    p = [0]*len(s)

    for i in range(len(p)):
        p[i] += int(s[i]) * t

    for i in range(len(p)-1):
        p[i+1] += (p[i] // 10)
        p[i] %= 10

    return list_to_str(([0]*k) + p)


def mul(str1, str2):
    lst1 = str_to_list(str1)
    lst2 = str_to_list(str2)

    neg_1 = is_neg(lst1)
    neg_2 = is_neg(lst2)

    if neg_1:
        lst1.pop()
    if neg_2:
        lst2.pop()

    result = "0"

    for i in range(len(lst2)):
        result = add(mul_ten(lst1, int(lst2[i]), i), result)

    if (neg_1 and (not neg_2)) or ((not neg_1) and neg_2):
        result = "-" + result

    return result

print(mul("+9", "-3"))

```
