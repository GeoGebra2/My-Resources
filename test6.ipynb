```python
import minimatrix as mm
m = 100
n = 10
X = mm.nrandom((m, n))
#print(f"X = {X}")
w = mm.nrandom((n, 1))
#print(f"w = {w}")
e = mm.nrandom((m, 1))#零均值？？？
summ = e.sum()
for i in range(e.dim[0]):
    for j in range(e.dim[1]):
        e.data[i][j] -= summ.data[0][0] / m
print(e.sum())
#print(f"e = {e}")
Y = X.dot(w) + e
#print(f"Y = X * w = {Y}")
X_T = X.T()
#print(f"X_T = {X_T}")
A = X_T.dot(X)
#print(f"A = X * XT = {A}")
inv_A = A.inverse()
#print(f"inverse A = {inv_A}")
B = inv_A.dot(X_T)
#print(f"B = {B}")
w_ = B.dot(Y)
print(w_)
print(w)
```
