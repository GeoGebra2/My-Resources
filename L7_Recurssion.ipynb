```python
def f(n):
    if n == 0:
        return n
    if n == 1:
        return 1

    return f(n - 1) + f(n - 2)


print(f(20))
```

```python
def factorial(n):
    print(f"{n} begins")
    if n == 0:
        return 1
    else:
        f = factorial(n - 1)
        print(f"{n-1} done")
        return n * f


print(factorial(5))
```

```python
def gcd(a, b):
    if a < b:
        return gcd(b, a)
    if b == 0:
        return a

    return gcd(b, a % b)


print(gcd(120, 88), gcd(90, 125))
```

```python
def sum_of_bits(n):
    return n if n == 0 or n == 1 else sum_of_bits(n // 2) + (n & 1)


print(sum_of_bits(7), sum_of_bits(5), sum_of_bits(4), sum_of_bits(1))
```

```python
def length_int(n):
    return 1 if n < 10 else length_int(n // 10) + 1


def inverse_int(n):
    return n if n < 10 else (n % 10) * 10 ** (length_int(n) - 1) + inverse_int(n // 10)


print(inverse_int(1001), inverse_int(1000), inverse_int(123456789))
```

```python
def my_pow(a, n):
    if n == 0:
        return 1

    t = my_pow(a, n // 2)

    return t * t if n % 2 == 0 else t * t * a


for i in range(4):
    print(my_pow(2, i))
```

```python
def Ackermann(m, n):
    if m == 0:
        return n + 1
    if n == 0:
        return Ackermann(m - 1, 1)

    return Ackermann(m - 1, Ackermann(m, n - 1))


print(Ackermann(3, 6))
print(Ackermann(4, 0))
```

```python
def sum_of_nested(number):
    if len(number) == 0:
        return 0

    if type(number[-1]) != type([]):
        return number[-1] + sum_of_nested(number[0:-1])

    return sum_of_nested(number[-1]) + sum_of_nested(number[0:-1])


number_lst = [[1, [1], [1]], [2], [3], 1, 1]
print(sum_of_nested(number_lst))

number_lst = [[[[[[1]]]]]]
print(sum_of_nested(number_lst))

number_lst = [[[[[[]]]]]]
print(sum_of_nested(number_lst))
```

```python
def binary_search(number, x, left, right):
    if left > right:
        return None

    mid = (left + right) // 2
    if number[mid] == x:
        return mid

    if number[mid] > x:
        return binary_search(number, x, left, mid - 1)

    return binary_search(number, x, mid + 1, right)


number = [-3, -4, 1, 3, 5]

print(binary_search(number, -6, 0, len(number) - 1))
print(binary_search(number, -3, 0, len(number) - 1))
print(binary_search(number, 1, 0, len(number) - 1))
print(binary_search(number, 0, 0, len(number) - 1))
print(binary_search(number, 5, 0, len(number) - 1))
print(binary_search(number, 7, 0, len(number) - 1))
```

```python
def perm(n):
    if n == 1:
        return [(1,)]

    lst = perm(n - 1)
    ans = []

    for x in lst:
        for i in range(n):
            nx = x[:i] + (n,) + x[i:]
            ans.append(nx)

    return ans


import pprint as pp

pp.pprint(perm(1))
pp.pprint(perm(2))
pp.pprint(perm(3))
pp.pprint(perm(4))
pp.pprint(perm(5))
pp.pprint(perm(6))
```

```python
def perm1(n):
    if n == 1:
        return [(1,)]

    lst = perm1(n - 1)
    ans = []

    for i in range(1, n + 1):  # 把元?? i 放到首位
        for x in lst:
            lx = list(x)

            for ii in range(n - 1):  # 用i+1,i+2,...,n代替原来排列??的i,i+1,...,n-1
                if lx[ii] >= i:
                    lx[ii] += 1

            nx = (i,) + tuple(lx)
            ans.append(nx)

    return ans


import pprint as pp

pp.pprint("Increasing permuation:")
pp.pprint(perm1(1))
pp.pprint(perm1(2))
pp.pprint(perm1(3))
pp.pprint(perm1(4))
```

```python
from itertools import permutations
perm = permutations(list(range(8)))
ans = 0

def valid(sln):
    for i in range(8):
        for j in range(i+1, 8):
            if  abs(sln[i]-sln[j]) == abs(i-j):
                return False
    return True

for x in perm:
    if valid(x):
        ans += 1

print(f"The number of solutions is {ans}")
```

```python
def Hanoi(n, x, y, z):
    if n == 1:
        print(f"{n}: {x}->{z}")
        return

    Hanoi(n - 1, x, z, y)
    print(f"{n}: {x}->{z}")
    Hanoi(n - 1, y, x, z)


print("Hanoi: n = 1")
Hanoi(1, 1, 2, 3)

print("Hanoi: n = 2")
Hanoi(2, 1, 2, 3)

print("Hanoi: n = 3")
Hanoi(3, 1, 2, 3)
```

```python
def digit_sum(x):
    ans = 0
    for i in str(x):
        ans += int(i)
    return ans


def digit_sum_re(x):
    if x < 10:
        return x

    return x % 10 + digit_sum_re(x // 10)


def digit_sum_while(x):
    ans = 0
    while x > 0:
        ans += x % 10
        x //= 10

    return ans


print(digit_sum(123), digit_sum_re(123), digit_sum_while(123))
```

```python
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    return fib(n - 1) + fib(n - 2)


def fib_loop_1(n):
    lst = [0] * (n + 1)
    lst[1] = 1
    for i in range(2, n + 1):
        lst[i] = lst[i - 1] + lst[i - 2]

    return lst[n]


def fib_loop_2(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    x1, x2 = 0, 1
    ans = 0
    for i in range(2, n + 1):
        ans = x1 + x2
        x1, x2 = x2, ans

    return ans


print(fib(20), fib_loop_1(20), fib_loop_2(20))
```
